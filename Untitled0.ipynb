Voici un exemple de code générique pour préparer $X$ et $Y$, diviser les données en entraînement/test, et appliquer un modèle d'apprentissage supervisé (Random Forest). Vous devrez remplacer `'NOM_DE_VOTRE_COLONNE_CIBLE'` par le nom réel de la colonne $Y$ que vous souhaitez prédire.

```python
import pandas as pd
from sklearn.model_selection import train_test_split
from sklearn.ensemble import RandomForestRegressor, RandomForestClassifier
from sklearn.preprocessing import LabelEncoder

# 1. Chargement robuste des données
try:
    df = pd.read_csv('/content/DiModica_AJH2022.csv', sep=None, engine='python', on_bad_lines='skip')
    print("Données chargées avec succès !")
    print(f"Dimensions : {df.shape[0]} lignes, {df.shape[1]} colonnes\n")
except Exception as e:
    print(f"Erreur lors de la lecture du fichier : {e}")
    df = None

if df is not None:
    # 2. Définition automatique de la variable cible Y (la dernière colonne par défaut)
    target_column = df.columns[-1]
    print(f"Variable cible (Y) : {target_column}")

    # Nettoyage rapide : suppression des lignes ayant des valeurs manquantes
    df_clean = df.dropna()

    # Séparation des caractéristiques (X) et de la cible (Y)
    X = df_clean.drop(columns=[target_column])
    Y = df_clean[target_column]

    # Encodage des variables textuelles/catégorielles en nombres pour X
    for col in X.select_dtypes(include=['object']).columns:
        le = LabelEncoder()
        X[col] = le.fit_transform(X[col].astype(str))

    # Encodage de Y si elle est textuelle (Classification)
    if Y.dtype == 'object':
        le_y = LabelEncoder()
        Y = le_y.fit_transform(Y.astype(str))
        is_classification = True
    else:
        is_classification = False

    # 3. Division des données en Train (80%) et Test (20%)
    X_train, X_test, Y_train, Y_test = train_test_split(X, Y, test_size=0.2, random_state=42)

    # 4. Choix et entraînement du modèle
    if is_classification:
        model = RandomForestClassifier(random_state=42)
        print("Modèle choisi : Random Forest Classifier (Classification)")
    else:
        model = RandomForestRegressor(random_state=42)
        print("Modèle choisi : Random Forest Regressor (Régression)")

    model.fit(X_train, Y_train)

    # 5. Évaluation des performances
    score = model.score(X_test, Y_test)
    print(f"Score de performance (R² ou Exactitude/Accuracy) : {score:.4f}")
```
